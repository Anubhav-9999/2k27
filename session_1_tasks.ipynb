### Q4:- Write a program to find the euclidean distance between two coordinates.Take both the coordinates from the user as input.

````python
import math

# Take inputs for the first coordinate (x1, y1)
x1 = float(input("Enter x1: "))
y1 = float(input("Enter y1: "))

# Take inputs for the second coordinate (x2, y2)
x2 = float(input("Enter x2: "))
y2 = float(input("Enter y2: "))

# Calculate the Euclidean distance
distance = math.sqrt((x2 - x1) ** 2 + (y2 - y1) ** 2)

# Display the result
print(f"\nThe Euclidean distance between ({x1}, {y1}) and ({x2}, {y2}) is: {distance:.2f}")
````

### Q6:- Write a program that will tell the number of dogs and chicken are there when the user will provide the value of total heads and legs.

For example:
Input:
heads -> 4
legs -> 12
<br>
Output:
dogs -> 2
chicken -> 2





````python
input_heads = int(input("enter total number of heads:"))
input_legs = int(input("enter total number of legs:"))
dogs = (input_legs - 2*input_heads)/2
chicken = input_heads - dogs
print(f"dogs = {dogs}, chicken = {chicken}")
````

### Q9:- Given 2 fractions, find the sum of those 2 fractions.Take the numerator and denominator values of the fractions from the user.

````python
import math

# Take inputs for the first fraction
num1 = int(input("Enter the numerator of the 1st fraction: "))
den1 = int(input("Enter the denominator of the 1st fraction: "))

# Take inputs for the second fraction
num2 = int(input("Enter the numerator of the 2nd fraction: "))
den2 = int(input("Enter the denominator of the 2nd fraction: "))

# Calculate the sum numerator and denominator
sum_num = (num1 * den2) + (num2 * den1)
sum_den = den1 * den2

# Simplify the fraction using the Greatest Common Divisor (GCD)
common_divisor = math.gcd(sum_num, sum_den)
simplified_num = sum_num // common_divisor
simplified_den = sum_den // common_divisor

# Display the result
print(f"\nResult: {num1}/{den1} + {num2}/{den2} = {sum_num}/{sum_den}")

# If it can be simplified, show the reduced version
if common_divisor > 1:
    print(f"Simplified form: {simplified_num}/{simplified_den}")
````

### Q10:- Given the height, width and breadth of a milk tank, you have to find out how many glasses of milk can be obtained? Assume all the inputs are provided by the user.



Input:<br>
Dimensions of the milk tank<br>
H = 20cm, L = 20cm, B = 20cm
<br><br>
Dimensions of the glass<br>
h = 3cm, r = 1cm

````python
import math

# Take input for the milk tank dimensions (in cm)
H = float(input("Enter tank height (H): "))
L = float(input("Enter tank length (L): "))
B = float(input("Enter tank breadth (B): "))

# Take input for the glass dimensions (in cm)
h = float(input("Enter glass height (h): "))
r = float(input("Enter glass radius (r): "))

# Calculate volumes
volume_tank = H * L * B
volume_glass = math.pi * (r ** 2) * h

# Calculate how many full glasses can be obtained
total_glasses = volume_tank / volume_glass
full_glasses = int(total_glasses)

# Display the results
print(f"\nVolume of the tank: {volume_tank:.2f} cubic cm")
print(f"Volume of one glass: {volume_glass:.2f} cubic cm")
print(f"Total full glasses of milk that can be obtained: {full_glasses}")
````
